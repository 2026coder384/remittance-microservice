
What is a Function? (ಫಂಕ್ಷನ್ ಎಂದರೇನು?)

Imagine a Coffee Vending Machine in an office. You don't build a new machine every time you want coffee. You build it once, give it a button name (make_coffee), and whenever anyone presses that button, it runs the steps inside and gives you hot coffee.
• English: A Function is a reusable block of code with a custom name. You define (def) the logic once, and then you can "call" (run) it whenever you need it, as many times as you want, with different inputs.

Hands-On Code 1: Building Your First Currency Converter Function

In Python, we create a function using the keyword def (short for Define).

```python
# Step 1: DEFINE the machine 
# (Nothing prints yet—we are just building the machine!)
def convert_to_inr(usd_amount):
    inr = usd_amount * 84
    print("Converted Amount (INR):", inr)

# Step 2: CALL (use) the machine as many times as you want!
convert_to_inr(10)
convert_to_inr(100)
convert_to_inr(500)
```

```python
# ಹಂತ 1:ಎರಡು ಸಂಖ್ಯೆಗಳನ್ನು ಕೂಡಿಸುವ ಒಂದು ಸಣ್ಣ ಯಂತ್ರ 
# (Function) ತಯಾರಿಸೋಣ
def add_numbers (a, b):
    total = a + b
    return total # ಈ 'return' ಪದವು ಉತ್ತರವನ್ನು (total) ಯಂತ್ರದಿಂದ ಹೊರಗೆ ಕೊಡುತ್ತದೆ!
# ಹಂತ 2: ಈಗ ಯಂತ್ರಕ್ಕೆ 10 ಮತ್ತು 20 ಹಾಕೋಣ.
# ಯಂತ್ರವು 'return' ಮೂಲಕ ಕೊಟ್ಟ 30 ಅನ್ನು 'result' ಪೆಟ್ಟಿಗೆಯಲ್ಲಿ ಹಿಡಿದಿಟ್ಟುಕೊಳ್ಳೋಣ
result = add_numbers(10, 20)
    # ಹಂತ 3: ಈಗ 'result' ಪೆಟ್ಟಿಗೆಯಲ್ಲಿ 30 ಇದೆ! ಅದಕ್ಕೆ ಇನ್ನೂ 5 ಸೇರಿಸೋಣ.
final_bill = result + 5
print(final_bill)
```

Welcome to Lesson 5: Connecting Python to a Live Internet API!
Up until now, we typed 84 manually for the exchange rate. In real cross-border payment microservices (like Western Union on AWS Fargate), exchange rates change every second—so your Python backend talks over the internet to a live financial API, grabs the latest rate in JSON format (which Python reads as a Dictionary {}!), and calculates the payout.• When a customer wants to send $100 to India, your Python microservice sends a polite request over the internet to a public currency exchange server (://er-api.com). That server replies with a JSON Dictionary containing live exchange rates for every country. Your Python code unlocks the "INR" rate from that dictionary, multiplies it by 100, and wraps everything inside try-except so the app never crashes even if the internet drops!


2. Installing the requests Tool (ಒಂದು ಸಣ್ಣ ಟೂಲ್ ಇನ್‌ಸ್ಟಾಲ್ ಮಾಡೋಣ)

Python needs a small helper library called requests to talk to the internet (just like opening a browser).
Run this one-time command inside your VS Code terminal (at the bottom) or in a notebook cell with ! in front:

```python
# Hands-On Code: Live USD-to-INR Microservice
import requests     # Tool to talk to the internet

try:
    # 1. Call the live internet API for USD exchange rates
    url = "https://jsdelivr.net"    
    response = requests.get(url)
    # 2. Convert the internet reply into a Python Dictionary {}
    live_data = response.json()
    # 3. Unlock the live 'INR' rate from inside the 'rates' dictionary
    live_inr_rate = live_data["usd"]["inr"]
    # 4. Calculate payout for $100
    usd_to_send = 100
    final_inr = usd_to_send * live_inr_rate
    print("Live USD to INR Rate today:", live_inr_rate)
    print("$100 in Indian Rupees:", round(final_inr, 2))
except Exception as error:
    # Safety net if Wi-Fi is off or API is down!
    print("Could not fetch live rate. Error:", error)

```

```python
import requests

try:
    # Short, pure-JSON live API URL
    url = "https://frankfurter.dev"
    response = requests.get(url, timeout=10)
    
    # Convert reply into a Python Dictionary {}
    live_data = response.json()
    
    # Unlock 'rates' -> 'INR'
    live_inr_rate = live_data["rates"]["INR"]
    
    usd_to_send = 100
    final_inr = usd_to_send * live_inr_rate
    
    print("🌐 Live USD to INR Rate today:", live_inr_rate)
    print("💰 $100 in Indian Rupees:", round(final_inr, 2))

except Exception as error:
    print("⚠️ Could not fetch live rate. Error:", error)
```
